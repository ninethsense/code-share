### Basic Flask API example
# pip install flask-api

```python
# Basic Flask API example
# Example: http://127.0.0.1:5000/hello

from flask import Flask 

app = Flask(__name__) # Initialize Flask app

@app.route('/hello/') # Define route for /hello/
def hello_world(): # Define function to handle requests to /hello/
    return {'message': 'Hello, World!'} # Return a JSON response

if __name__ == '__main__':
    app.run(debug=True, port=5000)
```

### Flask API to GET value from query string

```python
# Flask API with query parameter
# Example: http://127.0.0.1:5000/hello/?name=Praveen
from flask import Flask, request 

app = Flask(__name__) 

@app.route("/hello/", methods=['GET'], defaults={'name': 'No Name'}) # Default name if not provided
def hello_world(name): 
    name =  request.args.get('name') # Get the 'name' query parameter
    return {'message': f'Hello, World! {name}'}

if __name__ == '__main__':
    app.run(debug=True, port=5000)
```

### Flask API with GET and POST methods

```python
# Flask API with GET and POST methods
from flask import Flask, request 
app = Flask(__name__)

@app.route("/hello/", methods=['GET', 'POST'], defaults={'name': 'No Name'})
def hello_world(name):
    if request.method == 'POST':
        name = request.get_json().get('name', 'No Name')
    else:
        name = request.args.get('name')
    return {'message': f'Hello, World! {name}'}

if __name__ == '__main__':
    app.run(debug=True, port=5000)
```
