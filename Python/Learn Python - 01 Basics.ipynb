### Find the version of Python, installed

```python
# Know the version of Python you are using
import sys # import sys module to access system-specific parameters and functions
print("Python version: " + sys.version)
```

### Finding the type of the python Object

```python
# Find type of the object - use type() function
print(type(100)) # integer
print(type(100.34)) # float
print(type("Hello World")) # string
print(type({'Name': 'Alice', 'Age': 25})) # dictionary
print(type([1, 2, 3, 4, 5])) # list
print(type((10, 20, 30))) # tuple
print(type({1, 2, 3, 4})) # set
print(type(True)) # boolean
print(type(None)) # NoneType
print(type(lambda x: x + 1)) # function
print(type(range(5))) # range
print(type(b'Hello')) # bytes
print(type(bytearray(5))) # bytearray


```

### Convert one object type to another type

```python
# Convert one object type to another type
print(int(10.99))          # float to int
print(float(10))           # int to float
print(str(100))            # int to str
print(list((1, 2, 3)))     # tuple to list
print(tuple([4, 5, 6]))    # list to tuple
print(set([1, 2, 2, 3]))   # list to set
print(dict([('a', 1), ('b', 2)]))  # list of tuples to dict
print(bool(0))             # int to bool
print(bool(42))            # int to bool
print(bytes("Hello", 'utf-8'))  # str to bytes
print(bytearray(b'Hello'))  # bytes to bytearray
```

### Mathematical Operations

```python
print (10+20)  # Addition
print (30-15)  # Subtraction
print (5*6)    # Multiplication
print (40/8)   # Division
print (10//3)  # Floor/Integer Division
print (10%3)   # Modulus
print (2**3)   # Exponentiation
print (10 + 20 * 3)  # Operator Precedence
print ((10 + 20) * 3)  # Using Parentheses to change precedence
```

### Assignment Operators

```python
# Compound Assignment Operators
x = 10
x += 5  # x = x + 5
print (x)

# Multiplication Assignment Operator
x *= 2  # x = x * 2
print (x)

# Subtraction Assignment Operator
x -= 10 # x = x - 10
print (x)

# Division Assignment Operator
x /= 5  # x = x / 5
print (x)

# Modulus Assignment Operator
x %= 3  # x = x % 3
print (x)

# Exponentiation Assignment Operator
x **= 2 # x = x ** 2
print (x)
```

### Comparison Operators

```python
a = 10
b = 20
print(a == b)  # Equal to
print(a != b)  # Not equal to
print(a > b)   # Greater than
print(a < b)   # Less than
print(a >= b)  # Greater than or equal to
print(a <= b)  # Less than or equal to
```

### Strings

```python
str = "Praveen Nair"
print(str)
print(str[0])  # Accessing first character
print(str[-1]) # Accessing last character
print(str[0:7]) # Slicing from index 0 to 6
print(str[7:])  # Slicing from index 7 to end
print(str + " is a Man") # String concatenation
print(str[::2]) # Slicing with step
print(len(str)) # Length of the string
print(str.upper())  # Convert to uppercase
print(str.lower())  # Convert to lowercase
print(str.replace("Nair", "Kumar"))  # Replace substring
print(str.find("Nair"))  # Find substring
print(str.count("a"))  # Count occurrences of substring
print(str.startswith("Praveen"))  # Check if string starts with substring
print(str.endswith("Nair"))    # Check if string ends with substring    
print(str.split(" "))  # Split string into list
print (3 * "Hello ")  # String repetition
```

### String Formatting

```python
str = "is a string"
num = 10 
print(f"This {str} with number {num}") # f-string formatting
print("This {} with number {}".format(str, num)) # str.format() method
print("This %s with number %d" % (str, num)) # Old-style string formatting using % operator
print(f"Sum of 10 and 20 is {10 + 20}") # f-string with expression evaluation
print(r"regular string Hel\tlo\nWorld")  # Raw string literal
```
