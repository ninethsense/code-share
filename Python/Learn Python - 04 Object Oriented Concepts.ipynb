### Objects

```python
class Vehicle:
    def __init__(self, make, model, year): 
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"Vehicle Info: {self.year} {self.make} {self.model}")

car = Vehicle("Toyota", "Camry", 2020) # create instance/object of Vehicle
bike = Vehicle("Honda", "CBR500R", 2021) # create another instance/object of Vehicle
train = Vehicle("Siemens", "Velaro", 2019) # create another instance/object of Vehicle

car.display_info()
bike.display_info()
train.display_info()
```

### Inheritence

```python
# Single Inheritance example

class Animal:
    def speak(self):
        print("Animal speaks")
class Dog(Animal): # Dog inherits from Animal
    def bark(self):
        print("Dog barks")
my_dog = Dog() # create instance of Dog
my_dog.speak() # inherited method
my_dog.bark() # Dog's own method
```

```python
# Multiple Inheritance example
# Duck class inherits from both Flyer and Swimmer

class Flyer:
    def fly(self):
        print("Flying")
class Swimmer:
    def swim(self):
        print("Swimming")
class Duck(Flyer, Swimmer): # Duck inherits from Flyer and Swimmer
    def quack(self):
        print("Quacking")
my_duck = Duck() # create instance of Duck
my_duck.fly() # Flyer method
my_duck.swim() # Swimmer method
my_duck.quack() # Duck's own method
```

```python
# Multilevel Inheritance example
# Vehicle -> Car -> SportsCar

class Vehicle:
    def start(self):
        print("Vehicle started")
class Car(Vehicle): # Car inherits from Vehicle
    def drive(self):
        print("Car is driving")
class SportsCar(Car): # SportsCar inherits from Car
    def turbo_boost(self):
        print("Turbo boost activated")
my_sports_car = SportsCar() # create instance of SportsCar
my_sports_car.start() # Vehicle method
my_sports_car.drive() # Car method
my_sports_car.turbo_boost() # SportsCar method
```

```python
# Hybrid Inheritance example
# Base -> A, B -> C
# C inherits from both A and B, which inherit from Base

class Base:
    def base_method(self):
        print("Base method")
class A(Base): # A inherits from Base
    def method_a(self):
        print("Method A")
class B(Base): # B inherits from Base
    def method_b(self):
        print("Method B")
class C(A, B): # C inherits from A and B
    def method_c(self):
        print("Method C")
my_c = C() # create instance of C
my_c.base_method() # Base method
my_c.method_a() # A method
my_c.method_b() # B method
my_c.method_c() # C method
```

```python
# Hierarchical Inheritance example
# Multiple subclasses inherit from a single superclass

class Parent:
    def parent_method(self):
        print("Parent method")
class Child1(Parent): # Child1 inherits from Parent
    def child1_method(self):
        print("Child1 method")
class Child2(Parent): # Child2 inherits from Parent
    def child2_method(self):
        print("Child2 method")
my_child1 = Child1() # create instance of Child1
my_child2 = Child2() # create instance of Child2
my_child1.parent_method() # Parent method
my_child1.child1_method() # Child1 method
my_child2.parent_method() # Parent method
my_child2.child2_method() # Child2 method
```

### Polymorphism

```python
# Basic polymorphism example
# Compile-time polymorphism using method overloading
# Here, we simulate method overloading using default arguments

class Calc:
    def add(self, *a):
        return sum(a)
    
calc = Calc()
print(calc.add(2, 3))      
print(calc.add(2, 3, 4))  
print(calc.add(1, 2, 3, 4, 5)) 

```

```python
# Basic polymorphism example using method overriding
# Run-time polymorphism

class Parent:
    def show(self): # method to be overridden
        print("Parent show method")
class Child(Parent): # Child inherits from Parent
    def show(self):
        print("Child show method")
my_child = Child() # create instance of Child
my_child.show() # calls Child's overridden method
my_parent = Parent() # create instance of Parent
my_parent.show() # calls Parent's method

```

```python
# Duck Typing example
# different classes with same method name 

class Cat:
    def sound(self): # method with same name
        print("Meow")
class Dog:
    def sound(self): # method with same name
        print("Woof")
def make_sound(animal): # function expecting any animal
    animal.sound() # calls sound method
cat = Cat()
dog = Dog() 
make_sound(cat) # Meow
make_sound(dog) # Woof
```

```python
# Operator overloading example
# Define a Point class that overloads the + operator to add two Point objects
 
class Point:
    def __init__(self, x, y): # initializer
        self.x = x
        self.y = y
    def __add__(self, other): # overload + operator
        return Point(self.x + other.x, self.y + other.y)
    def display(self): # method to display point
        print(f"Point({self.x}, {self.y})")
p1 = Point(2, 3) # create first point
p2 = Point(4, 5) # create second point
p3 = p1 + p2 # uses overloaded + operator
p3.display() # Point(6, 8)
```

### Encapsulation

```python
# Encapsulation example
# Using private variables and public methods to access them

class Encapsulated:
    def __init__(self):
        self.__private_var = 42  # private variable
    def get_private_var(self): # public method to access private variable
        return self.__private_var
    def set_private_var(self, value): # public method to modify private variable
        self.__private_var = value

encap = Encapsulated() # create instance of Encapsulated
print(encap.get_private_var()) # access private variable via public method

encap.set_private_var(100) # modify private variable via public method
print(encap.get_private_var()) # access modified private variable via public method
```

### Data Abstration

```python
# Data Abstraction example
# Hiding complex implementation details using abstract base class


from abc import ABC, abstractmethod 
class AbstractVehicle(ABC): # abstract base class
    @abstractmethod
    def start_engine(self): # abstract method
        pass # must be implemented by subclasses

class Car(AbstractVehicle): # Car inherits from AbstractVehicle
    def start_engine(self): # implement abstract method
        print("Car engine started")
        
my_car = Car() # create instance of Car
my_car.start_engine() # Car engine started

```
