## Programming Fundamentals

### Conditional Operators

```python
a = 1
b = 2

print ("a = b" , a == b) # Equality
print ("a != b" , a != b) # Inequality
print ("a > b" , a > b)  # Greater than
print ("a < b" , a < b)  # Less than
print ("a >= b" , a >= b) # Greater than or equal to
print ("a <= b" , a <= b) # Less than or equal to

# string
print ("A = B" , 'A' == 'B') # Equality
print ("A != B" , 'A' != 'B') # Inequality
print ("A > B" , 'A' > 'B')  # Greater than
print ("A < B" , 'A' < 'B')  # Less than
print ("A >= B" , 'A' >= 'B') # Greater than or equal
print ("A <= B" , 'A' <= 'B') # Less than or equal to
```

### Comparison

```python
# simple if statement
x = 10
y = 20
if x < y:
    print("x is less than y")    
```

```python
# using Else
x = 10
y = 20
if x > y:
    print("x is greater than y")
else:
    print("x is not greater than y")
```

```python
# using Elif
x = 10
y = 20
if x > y:
    print("x is greater than y")
elif x == y:
    print("x is equal to y")
else:
    print("x is not greater than y")
```

### Logical Operations

```python
# simple if statement with logical operator or
x = 10
y = 20
if x < y or x == y:
    print("x is less than or equal to y")    


# simple if statement with logical operator and
x = 10
y = 20
if x < y and x != y:
    print("x is less than y")    
```

### Loops

```python
# for loop

for i in range(5):
    print("Iteration:", i)

# for loop with list
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
for day in days:
    print("Day:", day)

# for loop with index
for i, day in enumerate(days):
    print(f"Day: {day}", "Index:", i)

# for loop with _
for _ in range(3): # when the loop variable is not used
    greet("Praveen")

# nested for loop
for i in range(3):
    for j in range(2):
        print(f"i: {i}, j: {j}")
```

```python
# while loop
count = 0
while count < 5:
    print("Count:", count)
    count += 1

# while loop with break
count = 0
while True:
    print("Count:", count)
    count += 1
    if count >= 5:
        break

# while loop with continue
count = 0
while count < 5:
    count += 1
    if count == 3:
        continue
    print("Count:", count)
```

### Functions

```python
# simple function
def simple_function():
    print("This is a simple function")

simple_function()

# function with parameters
def sq(n):
    return n*n

print("Factorial of 5 is", sq(5))

# function with default parameters
def greet(name="Guest"):
    print("Hello,", name) 

greet()
greet("Praveen")  
```

```python
for _ in range(3):
    greet("Praveen")
```

### Exception Handling

```python
# Types of Python exceptions:
1. ValueError # invalid value 
2. TypeError # invalid type
3. IndexError # invalid index
4. KeyError # invalid key
5. AttributeError # invalid attribute
6. ZeroDivisionError # division by zero
7. ImportError # invalid import
```

```python
# using try-except
try: # code that may raise an exception
    len(['A','B',abcd])
except: # generic exception
    print("Some error occurred")


# using try-except-finally
try:
    len(['A','B',abcd])
except:
    print("Some error occurred")
finally: # always executes
    print("Execution completed")

# using try-except-else

try:
    len(['A','B',abcd])
except: 
    print("Some error occurred")
else: # executes if no exception occurs
    print("No error occurred")
finally:
    print("Execution completed")
```

```python
# using try-except with generic exception details

try:
    len(['A','B',abcd])
except Exception as e:
    print("An error occurred:", e)
```

```python
# using try-ecept with specific exception details

try:
    len(['A','B',abcd])      
except NameError as ne:
    print("Name error occurred:", ne)
except Exception as e:
    print("An error occurred:", e)

# Another way 
try:
    len(['A','B',abcd])      
except (NameError, Exception) as e:
    print("Name error occurred:", e)
```
