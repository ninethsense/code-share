## Data Structures

### Lists

```python
lst = ["Hello World", 1,2,3.0]
print(lst)

print (lst[0])  # Accessing first element
print (lst[1:3])  # Slicing from index 1 to 2
print (len(lst))  # Length of the list

lst.append("New Element")  # Adding an element
print (lst)

lst2 = [4,5,6, lst,["A", "B"]] # Nested list
print (lst2)

lst.extend([7,8,9])  # Extending the list
print (lst)

lst.remove(2)  # Removing an element
print (lst)

lst[0] = "Modified Element"  # Modifying an element
print (lst)

#del(lst[1])  # Deleting an element
print (lst)

new_lst = 'A,B,C,D,E'.split(",")  # Splitting a string into a list
print (new_lst)

list_A = [1,2,3]
list_B = [4,5,6]
combined_list = list_A + list_B  # Concatenating lists
print (combined_list)

list_C = [1,2,3]
list_D = list_C # Copying list reference
list_D[0] = 100
print ("A:", list_C) # Both reflect the change
print ("B:", list_D) # Both reflect the change



list_E = list_A[:] # Creating a shallow copy (clone by value) of the list
print ("E:", list_E)

list_E[0] = 100
print ("After modifying E:", list_E) # E is changed
print ("Value of A:", list_A) # A remains unchanged

list_A.reverse()  # Reversing the list
print (list_A)
```

### Tuples

```python
t1 = (1,2,3,"Hello")  # Creating a tuple
print (t1)
```

```python
print(t1[0])  # Accessing first element
print(t1[1:3])  # Slicing from index 1 to 2
print(len(t1))  # Length of the tuple
print (t1, type(t1))  # Display tuple and its type
print(type(t1[3]))  # Accessing the fourth element of the tuple and displaying its type
```

```python
t2 = t1 + (4,5,6)  # Concatenating tuples
print (t2)

t3 = (3,1,4,2)
sorted_t3 = sorted(t3)
print (sorted_t3)  # Sorted list from tuple elements

t5 = (10, 11, (1,2,3), (4,5), ("Hello",(6,7)))  # Nested tuple
print (t5)
print (t5[2][1])  # Accessing element '2' from nested tuple
print (t5[4][1][0])  # Accessing element '6'

print("Sum: ", sum(t3)) # Sum of tuple elements
print ("Min: ", min(t3)) # Minimum element in tuple
print ("Max: ", max(t3)) # Maximum element in tuple
```

```python
x =[1]
x.append([2,3,4,5])
print (len(x))
```

### Dictionaries

```python
dict1 = {"Name": "Alice", "Age": 30} # Creating a dictionary
print (dict1)
print (dict1["Name"])  # Accessing value by key
print (len(dict1))  # Length of the dictionary

dict1["Age"] = 31  # Modifying value by key
print (dict1)

dict1["City"] = "New York"  # Adding a new key-value pair
print (dict1)

# del(dict1["Name"])  # Deleting a key-value pair
print (dict1)
# Iterating through dictionary keys and values
for key, value in dict1.items():
    print (key, ":", value)

print ("Keys:", dict1.keys())  # Displaying all keys
print ("Values:", dict1.values())  # Displaying all values
print ("Items:", dict1.items())  # Displaying all key-value pairs
print (isinstance(dict1, dict))  # Checking if dict1 is a dictionary

# Demonstrating dictionary with various data types as values
dict2 = {
    "Integer": 100,
    "Float": 10.5,
    "String": "Hello",
    "List": [1,2,3],
    "Tuple": (4,5,6),
    "NestedDict": {"Key1": "Value1", "Key2": "Value2"}
}
print (dict2)
print (dict2["List"][1])  # Accessing second element of the list in the dictionary
print (dict2["NestedDict"]["Key2"])  # Accessing value from nested dictionary
print (dict2)


dict3 = dict(Name="Bob", Age=25, City="Los Angeles")  # Creating a dictionary using dict()
print (dict3)




```

### Sets

```python
x={"a","a","c",}  # Creating a set
print (x) # Displaying the set (duplicates removed)

# Creating a list with duplicates and converting to set
dy=["a","a","c",] # Creating a list with duplicates
y = set(dy)
print (y) # Displaying the list (duplicates retained)
print (dy) # Displaying the list (duplicates retained)

x.add(5)  # Adding an element to the set
print (x)
x.remove(5) # Removing an element from the set
print (x)

# Set operations
set_A = {1,2,3,4}
set_B = {3,4,5,6}
print ("Union:", set_A | set_B)  # Union of sets
print ("Intersection:", set_A & set_B)  # Intersection of sets
print ("Difference (A-B):", set_A - set_B)  # Difference of sets
print ("Difference (B-A):", set_B - set_A)  # Difference of sets
print ("Symmetric Difference:", set_A ^ set_B)  # Symmetric difference of sets
print ("Is A subset of B?:", set_A.issubset(set_B))  # Subset check
print ("Is A superset of B?:", set_A.issuperset(set_B))  # Superset check
print ("Are A and B disjoint?:", set_A.isdisjoint(set_B))  # Disjoint check
print (isinstance(x, set))  # Checking if x is a set

print ("'a' in x :", "a" in x)  # Membership test

frozenset_A = frozenset([1,2,3,4])  # Creating a frozenset
print (frozenset_A)  # Displaying the frozenset



```
