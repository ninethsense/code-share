## What is Asyncio?

*asyncio* is a library to write concurrent code using the async/await syntax. asyncio means Asynchronous I/O
Read more: https://docs.python.org/3/library/asyncio.html

## Basic asyncio example

```python
# Basic asyncio example

import asyncio # import the asyncio module
async def main(): # define an asynchronous function
    print('Hello ...') 
    await asyncio.sleep(3) # simulate a non-blocking delay of 3 seconds
    print('... World!') # print after the delay

await main()
```

```python
# Another basic example 
# Using async/await to fetch data# 
async def read_data(): 
    print('Start fetching data...')
    await asyncio.sleep(2) # simulate a network delay
    print('Data fetched!')
    return {'data': 123}

async def main_fetch():
    data = await read_data() # wait for fetch_data to complete
    print(f'Received: {data}')

await main_fetch()
```

```python
import time # import the time module, just to check execution time

start_time = time.perf_counter() # record the start time

print('Before sleep...')
await asyncio.sleep(3)  # keep the event loop running for a bit
end_time = start_time - time.perf_counter()  # record the end time
print(f"After sleep...{end_time:.4f} seconds elapsed")

# Another example with result
await asyncio.sleep(5, result='Finished')  # sleep for 2 seconds and return 'Finished'

# Another example with a callback function
def finished():
    print("Sleep is over!")

await asyncio.sleep(5, result=finished())
```
