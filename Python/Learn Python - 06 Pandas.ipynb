### Data Frame

```python
# Basic pandas example
lst = [1, 2, 3, 4, 5]
df = pd.DataFrame(lst, columns=['Numbers'])
print(df)

# Another basic pandas example
# dictionary of planets data
planets_dict = {
    'name': ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune'],
    'mass': [0.330, 4.87, 5.97, 0.642, 1898, 568, 86.8, 102],
    'radius': [2439.7, 6051.8, 6371.0, 3389.5, 69911, 58232, 25362, 24622],
    'distance_from_sun': [57.9, 108.2, 149.6, 227.9, 778.6, 1433.5, 2872.5, 4495.1]
}

# create a DataFrame
planets_df = pd.DataFrame(planets_dict)
print(planets_df)
```

```python
print("planets_df['name']:", planets_df["name"])  # Print the 'name' column of the DataFrame
print("planets_df[['name', 'mass']]:", planets_df[['name', 'mass']])  # Print the 'name' and 'mass' columns of the DataFrame
```

```python
print (planets_df.head())  # Display the first few rows of the DataFrame
print ("\n",planets_df.tail())  # Display the first few rows of the DataFrame
```

```python
# Print summary statistics of the DataFrame
print("planets_df.describe():\n",planets_df.describe()) 
```

```python
print("planets_df.iloc[0]:\n",planets_df.iloc[0]) # Print the first row of the DataFrame
print("\nplanets_df.loc[0]:",planets_df.loc[0]) # Print the row with index label 0 of the DataFrame
```

```python
print(planets_df[['name', 'mass']])  # Print the 'name' and 'mass' columns of the DataFrame
print("\n", planets_df[0:2]) # Print the first two rows of the DataFrame
```

### Finding elements

### 

```python
print("Unique elements: ", planets_df['name'].unique())  # Get unique planet names
```

```python
# Filtering rows where the planet name is 'Earth'
df1 = planets_df["name"] == "Earth"
print("Planets named Earth:\n", planets_df[df1])  # Print rows where
```

```python
# Save data frame
planets_df.to_csv('planets_data.csv', index=False);
```
